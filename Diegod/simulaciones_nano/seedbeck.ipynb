# Simulaciones de funciones de fermi 

```python
### CÓDIGO PARA PLOTEAR BIEN LA FUNCIÓN DE FERMI #################


import numpy as np
import matplotlib.pyplot as plt

# =========================================================
# PARÁMETROS
# =========================================================

EF = 0.0          # Energía de Fermi en eV

T1 = 300          # Temperatura 1 en K
T2 = 10          # Temperatura 2 en K

kB = 8.617e-5     # Constante de Boltzmann en eV/K

Emin = -0.15       # Energía mínima en eV
Emax = 0.15        # Energía máxima en eV
num_puntos = 1000

# =========================================================
# FUNCIÓN DE FERMI
# =========================================================

def fermi(E, EF, T):
    return 1 / (1 + np.exp((E - EF) / (kB * T)))

# =========================================================
# EJE DE ENERGÍAS
# =========================================================

E = np.linspace(Emin, Emax, num_puntos)

# Calculamos las dos funciones de Fermi
f_T1 = fermi(E, EF, T1)
f_T2 = fermi(E, EF, T2)

# Diferencia entre ambas
diferencia = f_T1 - f_T2

# =========================================================
# PLOT 1: FUNCIONES DE FERMI
# =========================================================

plt.figure(figsize=(8, 5))

plt.plot(E, f_T1, label=rf"$T_1$ = {T1} K")
plt.plot(E, f_T2, label=rf"$T_2$ = {T2} K")

plt.axvline(EF, linestyle="--", color="black", label=r"$E_F$")

plt.xlabel("Energía E (eV)")
plt.ylabel(r"$f(E, E_F)$")
plt.title(rf"Funciones de Fermi a temperaturas $T_1$={T1} y $T_2$={T2}")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(f"Funciones de fermi a T1={T1} y T2={T2}.jpg")
plt.show()

# =========================================================
# PLOT 2: DIFERENCIA ENTRE FUNCIONES DE FERMI
# =========================================================

plt.figure(figsize=(8, 5))

plt.plot(E, diferencia, label=rf"$f_{{T_1={T1}K}} - f_{{T_2={T2}K}}$")

plt.axhline(0, linestyle="--", color="black")
plt.axvline(EF, linestyle="--", color="black", label=r"$E_F$")

plt.xlabel("Energía E (eV)")
plt.ylabel("Diferencia")
plt.title(rf"Diferencia entre funciones de Fermi a $T_1$={T1} y $T_2$={T2}")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(rf"Diferencia entre fermi a T1={T1} y T2={T2}.jpg")
plt.show()
```
