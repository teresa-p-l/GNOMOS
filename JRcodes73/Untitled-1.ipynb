# Estudiar tamaño del gap

Se puede estudiar el tamaño del gap teniendo en cuenta que es mucho mas grande que la diferencia entre puntos que no estan en el gap

```python
def find_gap(graph):
    gaps = []
    difs = np.diff(graph.iloc[:,1])
    for i in range(len(difs)-1):
        if difs[i] > 0.05:
            print(f'Gap found between {graph.iloc[i,0]} and {graph.iloc[i+1,0]} with a difference of {difs[i]}')
            gaps.append((graph.iloc[i,0], graph.iloc[i+1,0], difs[i]))
    return gaps

find_gap(b_08_w_04_cent)
```

```python
print("b_04_w_0:")
find_gap(b_04_w_0)
print("\nb_08_w_0:")
find_gap(b_08_w_0)
print("\nb_12_w_0:")
find_gap(b_12_w_0)
print("\nb_08_w_04:")
find_gap(b_08_w_04)
print("\nb_12_w_04:")
find_gap(b_12_w_04)
print("\nb_12_w_08:")
find_gap(b_12_w_08)
```
